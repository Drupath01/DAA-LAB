# 0/1 Knapsack Problem

Given weights and values of $n$ items, we need to put these items in a knapsack of capacity $W$ to get the maximum total value in the knapsack.

```python
def knapsack_recursive(W, wt, val, n, memo=None):
    if memo is None:
        memo = {}

    # Base Cases
    if n == 0 or W == 0:
        return 0

    # Check memo table
    if (n, W) in memo:
        return memo[(n, W)]

    # If weight of the nth item is more than Knapsack capacity W,
    # then this item cannot be included in the optimal solution
    if wt[n-1] > W:
        result = knapsack_recursive(W, wt, val, n-1, memo)
    else:
        # Return the maximum of two cases:
        # (1) nth item included
        # (2) not included
        result = max(
            val[n-1] + knapsack_recursive(W - wt[n-1], wt, val, n-1, memo),
            knapsack_recursive(W, wt, val, n-1, memo)
        )

    memo[(n, W)] = result
    return result

# Test data
val = [60, 100, 120]
wt = [10, 20, 30]
W = 50
n = len(val)
print(f"[Recursive with Memoization] Max Value: {knapsack_recursive(W, wt, val, n)}")
```

```python
def knapsack_iterative(W, wt, val, n):
    # Create a 2D table to store results of subproblems
    dp = [[0 for _ in range(W + 1)] for _ in range(n + 1)]

    # Build table dp[][] in bottom-up manner
    for i in range(n + 1):
        for w in range(W + 1):
            if i == 0 or w == 0:
                dp[i][w] = 0
            elif wt[i-1] <= w:
                dp[i][w] = max(val[i-1] + dp[i-1][w-wt[i-1]], dp[i-1][w])
            else:
                dp[i][w] = dp[i-1][w]

    return dp[n][W]

# Test data
val = [60, 100, 120]
wt = [10, 20, 30]
W = 50
n = len(val)
print(f"[Iterative Dynamic Programming] Max Value: {knapsack_iterative(W, wt, val, n)}")
```
