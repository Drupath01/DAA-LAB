# Coin Change Problem

Given an amount $V$ and an array of coin denominations `coins`, find the minimum number of coins required to make the change. If it's not possible to make that amount, return -1.

```python
def coin_change_recursive(coins, amount, memo=None):
    if memo is None:
        memo = {}

    # Base Cases
    if amount == 0:
        return 0
    if amount < 0:
        return -1
    if amount in memo:
        return memo[amount]

    min_coins = float('inf')

    for coin in coins:
        res = coin_change_recursive(coins, amount - coin, memo)
        if res >= 0 and res < min_coins:
            min_coins = res + 1

    memo[amount] = -1 if min_coins == float('inf') else min_coins
    return memo[amount]

# Test data
coins = [1, 2, 5]
amount = 11
print(f"[Recursive with Memoization] Min coins for {amount}: {coin_change_recursive(coins, amount)}")
```

```python
def coin_change_iterative(coins, amount):
    # dp[i] will store the minimum coins needed for amount i
    dp = [float('inf')] * (amount + 1)
    dp[0] = 0

    for i in range(1, amount + 1):
        for coin in coins:
            if i - coin >= 0:
                dp[i] = min(dp[i], dp[i - coin] + 1)

    return dp[amount] if dp[amount] != float('inf') else -1

# Test data
coins = [1, 2, 5]
amount = 11
print(f"[Iterative Dynamic Programming] Min coins for {amount}: {coin_change_iterative(coins, amount)}")
```
