# Matrix Chain Multiplication

Given a sequence of matrices, find the most efficient way to multiply these matrices together. The problem is not actually to perform the multiplications, but merely to decide in which order to perform the multiplications.

```python
import sys

def matrix_chain_recursive(p, i, j, memo=None):
    if memo is None:
        memo = {}

    # Base Case: one matrix
    if i == j:
        return 0

    if (i, j) in memo:
        return memo[(i, j)]

    minimum_computations = sys.maxsize

    # Place parenthesis at different places between first and last matrix
    for k in range(i, j):
        count = (
            matrix_chain_recursive(p, i, k, memo)
            + matrix_chain_recursive(p, k + 1, j, memo)
            + p[i-1] * p[k] * p[j]
        )
        if count < minimum_computations:
            minimum_computations = count

    memo[(i, j)] = minimum_computations
    return minimum_computations

# Test data: dimensions of matrices A_1 (10x20), A_2 (20x30), A_3 (30x40), A_4 (40x30)
arr = [10, 20, 30, 40, 30]
size = len(arr)
print(f"[Recursive with Memoization] Minimum multiplications: {matrix_chain_recursive(arr, 1, size - 1)}")
```

```python
import sys

def matrix_chain_iterative(p, n):
    # dp[i][j] stores the minimum number of multiplications needed to compute the matrix A[i]...A[j]
    dp = [[0 for _ in range(n)] for _ in range(n)]

    # l is chain length
    for l in range(2, n):
        for i in range(1, n - l + 1):
            j = i + l - 1
            dp[i][j] = sys.maxsize
            for k in range(i, j):
                q = dp[i][k] + dp[k + 1][j] + p[i - 1] * p[k] * p[j]
                if q < dp[i][j]:
                    dp[i][j] = q

    return dp[1][n - 1]

# Test data
arr = [10, 20, 30, 40, 30]
size = len(arr)
print(f"[Iterative Dynamic Programming] Minimum multiplications: {matrix_chain_iterative(arr, size)}")
```
