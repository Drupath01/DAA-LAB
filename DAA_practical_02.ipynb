## Search Algorithms: Linear Search and Binary Search

```python
import time
import random

def linear_search(arr, target):
    """Performs a linear search on a list.

    Args:
        arr: The list to search through.
        target: The element to search for.

    Returns:
        The index of the target if found, otherwise -1.
    """
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1

# Example usage and execution time measurement
print("--- Linear Search ---")
my_list = list(range(1, 1000000))
search_target = 999999 # Target near the end for worst case

start_time = time.time()
index = linear_search(my_list, search_target)
end_time = time.time()

if index != -1:
    print(f"Target {search_target} found at index: {index}")
else:
    print(f"Target {search_target} not found in the list.")

print(f"Execution time for Linear Search: {end_time - start_time:.6f} seconds")

# Example with target not found
search_target_not_found = 1000001
start_time = time.time()
index_not_found = linear_search(my_list, search_target_not_found)
end_time = time.time()

if index_not_found == -1:
    print(f"Target {search_target_not_found} not found (correctly).")
print(f"Execution time for Linear Search (not found): {end_time - start_time:.6f} seconds")
```

```python
import time

def binary_search(arr, target):
    """Performs an iterative binary search on a sorted list.

    Args:
        arr: The sorted list to search through.
        target: The element to search for.

    Returns:
        The index of the target if found, otherwise -1.
    """
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1

# Example usage and execution time measurement
print("\n--- Binary Search ---")
sorted_list = list(range(1, 1000000)) # Must be sorted!
search_target = 999999

start_time = time.time()
index = binary_search(sorted_list, search_target)
end_time = time.time()

if index != -1:
    print(f"Target {search_target} found at index: {index}")
else:
    print(f"Target {search_target} not found in the list.")

print(f"Execution time for Binary Search: {end_time - start_time:.6f} seconds")

# Example with target not found
search_target_not_found = 1000001
start_time = time.time()
index_not_found = binary_search(sorted_list, search_target_not_found)
end_time = time.time()

if index_not_found == -1:
    print(f"Target {search_target_not_found} not found (correctly).")
print(f"Execution time for Binary Search (not found): {end_time - start_time:.6f} seconds")
```
