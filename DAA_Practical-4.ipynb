# Recursion vs Iteration

Here we will compare recursion (a function calling itself) and iteration (using loops) using the classic example of calculating a factorial.

```python
def factorial_recursive(n):
    # Base case
    if n <= 1:
        return 1
    # Recursive step
    return n * factorial_recursive(n - 1)

# Demonstration
num = 5
print(f"[Recursion] Factorial of {num} is {factorial_recursive(num)}")
```

```python
def factorial_iterative(n):
    result = 1
    # Loop iteration
    for i in range(1, n + 1):
        result *= i
    return result

# Demonstration
num = 5
print(f"[Iteration] Factorial of {num} is {factorial_iterative(num)}")
```
